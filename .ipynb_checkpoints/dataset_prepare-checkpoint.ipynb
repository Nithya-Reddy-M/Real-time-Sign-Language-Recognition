```python
import os
import random
import shutil

VIDEO_PATH = "videos"

# Folder for selected videos
SELECTED_PATH = "selected_videos"

# Create folder
os.makedirs(SELECTED_PATH, exist_ok=True)

# Get only mp4 files
all_videos = [f for f in os.listdir(VIDEO_PATH) if f.endswith(".mp4")]

print("Total videos found:", len(all_videos))

# Shuffle randomly
random.shuffle(all_videos)

# Select first 60
selected_videos = all_videos[:60]

# Copy them
for vid in selected_videos:
    src = os.path.join(VIDEO_PATH, vid)
    dst = os.path.join(SELECTED_PATH, vid)
    shutil.copy(src, dst)

print("Selected videos:", len(selected_videos))
```
