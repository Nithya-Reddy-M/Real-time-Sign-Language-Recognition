```python
import json
import os
import yt_dlp

# Load JSON file
with open('WLASL_v0.3.json') as f:
    data = json.load(f)

# Create folder to store videos
save_path = "videos"
os.makedirs(save_path, exist_ok=True)

ydl_opts = {
    'outtmpl': f'{save_path}/%(id)s.%(ext)s',
    'quiet': True
}

# Download limited signs (start small)
for entry in data[:50]:   # change to 100 later if needed
    for instance in entry['instances']:
        url = instance['url']
        try:
            with yt_dlp.YoutubeDL(ydl_opts) as ydl:
                ydl.download([url])
        except:
            pass

print("Download complete")
```
