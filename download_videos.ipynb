```python
import json
import os
import yt_dlp

# Load JSON
with open('WLASL_v0.3.json') as f:
    data = json.load(f)

# Create folder
save_path = "videos"
os.makedirs(save_path, exist_ok=True)

ydl_opts = {
    'outtmpl': f'{save_path}/%(id)s.%(ext)s',
    'quiet': True,
    'nocheckcertificate': True,   # 🔥 fixes SSL issue
    'ignoreerrors': True          # 🔥 skip broken videos
}

count = 0

for entry in data[:50]:
    for instance in entry['instances']:
        url = instance['url']

        if "youtube.com" not in url:
            continue   # skip non-youtube links

        try:
            with yt_dlp.YoutubeDL(ydl_opts) as ydl:
                ydl.download([url])
                count += 1
        except:
            pass

print(f"Downloaded {count} videos")
```
